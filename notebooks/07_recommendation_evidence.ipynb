# Final recommendation evidence

Corrected Idea 1 and the rebuilt cart-completion proposal use the verified shared dataset. Set private input paths before starting the kernel. The earned-reward option is deferred. This notebook does not access an online report or launch a campaign. Keep executed copies and results outside Git.

```python
from pathlib import Path
import importlib.util
import json
import os
import sys

cwd = Path.cwd().resolve()
repo = next((p for p in [cwd, *cwd.parents] if (p / 'scripts' / 'build_recommendation_evidence.py').is_file()), None)
if repo is None:
    raise RuntimeError('Start Jupyter inside this repository.')
sys.path.insert(0, str(repo / 'src'))
spec = importlib.util.spec_from_file_location('evidence_runner', repo / 'scripts' / 'build_recommendation_evidence.py')
runner = importlib.util.module_from_spec(spec)
spec.loader.exec_module(runner)
report = runner.build_evidence(
    welcome_codes_path=os.environ.get('WEAVVE_WELCOME_CODES'),
    post_purchase_scenario_path=os.environ.get('WEAVVE_IDEA1_SCENARIO'),
    cart_scenario_path=os.environ.get('WEAVVE_IDEA2_SCENARIO'),
    reward_scenario_path=os.environ.get('WEAVVE_DEFERRED_REWARD_SCENARIO'),
    ads_dir=os.environ.get('WEAVVE_ADS_SNAPSHOTS'),
)
print('Dataset:', report['idea1']['dataset_id'])
print('Cutoff:', report['idea1']['transaction_cutoff'])
```
