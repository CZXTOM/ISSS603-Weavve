# 04 Build the shared cleaned dataset

Run this notebook first, from the `notebooks` folder. Set `WEAVVE_DATA` to the supplied raw files or ZIPs and `WEAVVE_OUT` to a private output folder outside the Git repository **before starting the kernel**.

One build creates the canonical transaction, customer mapping, product, promotion and search tables, a data dictionary, a manifest and a local ZIP. The manifest records the raw-source hashes, cleaning rules, code hashes and output schemas. Notebooks 05 and 10–13 read this same dataset. Matching dataset IDs identify the same source/code/rule contract.

Review the flags below before analysing the tables. Building and validating a dataset does not prove causal marketing effects or the report's conclusions. The generated files and executed notebook outputs contain private data; keep them outside GitHub.


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent / "src"))

import pandas as pd
from IPython.display import display
from weavve.clean import build_dataset

dataset = build_dataset(make_zip=True)
print("Shared dataset ID:", dataset.manifest["dataset_id"])
print("Local dataset:", dataset.path)
print("Local ZIP:", dataset.path.parent / f"weavve_clean_{dataset.manifest['dataset_id'][:16]}.zip")
print("Transaction cutoff:", dataset.manifest["transaction_cutoff"])
```

## Inspect the output inventory

Counts describe table grains, not unique customers or independent observations. For example, voucher rows are issued rewards, subscriber rows are snapshot profiles, and search candidate rows retain competing exports for review.


```python
inventory = pd.DataFrame([
    {"table": name, "rows": entry["rows"], "columns": len(entry["columns"])}
    for name, entry in sorted(dataset.manifest["tables"].items())
])
display(inventory)
display(pd.Series(dataset.manifest["checks"]["transactions"], name="value").to_frame())
```

## Review unresolved source meanings

These flags travel with the shared dataset. They must remain visible when presenting findings. The manifest and `DATA_DICTIONARY.md` are in the local output directory printed above.


```python
for warning in dataset.manifest["warnings"]:
    print("-", warning)
if not dataset.manifest["warnings"]:
    print("No automated review flags were emitted; source meaning still requires review.")
```
